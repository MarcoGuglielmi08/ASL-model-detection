# Engineered Features Dataset Analysis - ASL

This notebook analyzes `asl_features_engineered.csv`, the tabular dataset with engineered geometric features derived from hand landmarks.

Goal: inspect data quality and feature behavior before model training.

## Imports and Configuration

```python
from pathlib import Path
from displayfunction import display

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from util.paths import DATA_DIR

plt.rcParams["figure.figsize"] = (12, 7)
pd.set_option("display.max_columns", 140)
```

```python
def find_project_root():
    current = Path.cwd().resolve()
    for candidate in [current, *current.parents]:
        if (candidate / "data" / "asl_features_engineered.csv").exists():
            return candidate
    raise FileNotFoundError("Could not find data/asl_features_engineered.csv starting from the current folder")


PROJECT_ROOT = find_project_root()
FEATURES_CSV = DATA_DIR / "asl_features_engineered.csv"
FEATURES_CSV
```

## CSV Loading

```python
df = pd.read_csv(FEATURES_CSV)

print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns):,}")
df.head()
```

```python
metadata_cols = [c for c in ["filepath", "label", "handedness"] if c in df.columns]
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

print("Metadata columns:", metadata_cols)
print("Numeric features:", len(numeric_cols))

prefix_counts = pd.Series(df.columns).map(lambda c: c.split("_")[0]).value_counts()
prefix_counts
```

## Basic Data Quality Checks

```python
missing_total = int(df.isna().sum().sum())
duplicate_filepaths = int(df["filepath"].duplicated().sum()) if "filepath" in df.columns else 0
duplicate_rows = int(df.duplicated().sum())

print(f"Total missing values: {missing_total}")
print(f"Duplicated filepaths: {duplicate_filepaths}")
print(f"Duplicated full rows: {duplicate_rows}")
```

```python
if "label" in df.columns:
    label_counts = df["label"].value_counts().sort_index()
    label_counts_df = label_counts.rename_axis("label").reset_index(name="count")
    label_counts_df["percent"] = label_counts_df["count"] / len(df) * 100
    display(label_counts_df)

    ax = label_counts.plot(kind="bar")
    ax.set_title("Examples per Class (Engineered Features)")
    ax.set_xlabel("Label")
    ax.set_ylabel("Count")
    plt.xticks(rotation=0)
    plt.show()
else:
    print("Column 'label' not found.")
```

```python


if "handedness" in df.columns:
    handedness_counts = df["handedness"].value_counts(dropna=False)
    display(handedness_counts)

    ax = handedness_counts.plot(kind="bar")
    ax.set_title("Handedness Distribution")
    ax.set_xlabel("Handedness")
    ax.set_ylabel("Count")
    plt.xticks(rotation=0)
    plt.show()
else:
    print("Column 'handedness' not found.")
```

## Feature Scale and Variability

```python
feature_summary = pd.DataFrame({
    "min": df[numeric_cols].min(),
    "mean": df[numeric_cols].mean(),
    "std": df[numeric_cols].std(),
    "max": df[numeric_cols].max(),
})

feature_summary.head(20)
```

```python
top_variance = df[numeric_cols].var().sort_values(ascending=False).head(20)
top_variance
```

```python
top_var_cols = top_variance.index.tolist()[:8]

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for ax, col in zip(axes.flat, top_var_cols):
    df[col].plot(kind="hist", bins=50, ax=ax)
    ax.set_title(col)

plt.tight_layout()
plt.show()
```

## Simple Outlier Screening

```python
df_outlier = df.copy()
df_outlier["max_abs_feature"] = df_outlier[numeric_cols].abs().max(axis=1)

cols_to_show = [c for c in ["filepath", "label", "handedness", "max_abs_feature"] if c in df_outlier.columns]
df_outlier[cols_to_show].sort_values("max_abs_feature", ascending=False).head(20)
```

## Feature/Target Matrix for Training

```python
X = df[numeric_cols].astype("float32")
y = df["label"] if "label" in df.columns else None

print("X shape:", X.shape)
if y is not None:
    print("y shape:", y.shape)
    print("Classes:", sorted(y.unique()))
```

## Final Notes

This dataset contains raw landmark-derived features (`lm_*`) and additional geometric descriptors (distances, angles, vectors) that can help improve class separability.

Before training, consider scaling numeric features (for example with StandardScaler) and using a stratified train/validation split by `label`.
